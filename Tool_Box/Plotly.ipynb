```python
fig = go.Figure()
fig.add_trace(go.Scatter(x=[1, 2, 3, 4, 5], y=[3, 4, 5, 7, 8], mode='markers'))
fig.show()
```

```python
fig = go.Figure()
fig.add_trace(go.Scatter(x=[1, 2, 3, 4, 5], y=[3, 4, 5, 7, 8], mode='lines'))
fig.show()
```

```python
fig = go.Figure()
fig.add_trace(go.Bar(x=[1, 2, 3, 4, 5], y=[3, 4, 5, 7, 8]))
fig.show()
```

```python
x = [1,2,3,4,5,4,13,11,1,5,6,12,12]
fig =go.Figure(data=[go.Histogram(x = x)])
fig.show()
```

```python
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
```

```python
fig = go.Figure(data = [go.Histogram(x = tips.total_bill)])
fig.show()
```

```python
fig = go.Figure()
fig.add_trace(go.Scatter(x=tips.total_bill, y=tips.tip, mode='markers'))
fig.show()
```

```python
fig = go.Figure()
fig.add_trace(go.Scatter(x=tips.total_bill, y=tips.tip, mode='markers', marker_size=10*tips['size']))
fig.show()
```

```python
fig = go.Figure()
fig.add_trace(go.Scatter3d(x=tips.total_bill, y=tips.tip, mode='markers', z=tips['size']))
fig.show()
```

```python
fig = go.Figure()
fig.add_trace(go.Scatter3d(x=tips.total_bill, y=tips.tip, mode='lines', z=tips['size']))
fig.show()
```
