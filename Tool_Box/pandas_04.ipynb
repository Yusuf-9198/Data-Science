```python
data1 = pd.DataFrame({'key1':[1,2,4,6,8,3],
                      'key2':[3,2,46,8,6,76],
                      'key3':[2,8,54,222,56,9]
                      })
```

```python
data2 = pd.DataFrame({'key1': [1, 2, 44, 6, 8, 3],
                      'key2': [3, 2, 46, 87, 67,76],
                      'key3': [2, 81, 54, 22, 56, 9],
                      'key4':[76,43,78,98,60,66]
                      })
```

```python
data1
```

```python
data2
```

```python
# merge operation
```

```python
pd.merge(data1,data2) # inner join karta hai
```

```python
pd.merge(data1,data2,how='left') # left join
```

```python
pd.merge(data1,data2,how='right') # right join
```

```python
pd.merge(data1,data2,how='outer',on='key2') # outer join,key2 common lo
```

```python
pd.merge(data1,data2,how='cross') # cross join(cartitian product
```
